```python
# Setup
import seaborn as sns

df = sns.load_dataset("titanic")
```

## Exercises

1. How many passengers and how many columns are in the dataset?
2. What is the average fare (rounded to 2 decimals)?
3. How many passengers were children (age under 18)?
4. What percentage of 3rd-class passengers survived?
5. Show age, sex, class and survived for the 5 oldest passengers.

```python
# Solution 1
df.shape
```

```python
# Solution 2
round(df["fare"].mean(), 2)
```

```python
# Solution 3
len(df[df["age"] < 18])        # or: (df["age"] < 18).sum()
```

```python
# Solution 4
df[df["pclass"] == 3]["survived"].mean() * 100
```

```python
# Solution 5
df.sort_values("age", ascending=False)[["age", "sex", "pclass", "survived"]].head()
```

## Practice
1. How many passengers embarked at each town (`embark_town`)?
2. Survival rate of men vs. women.
3. Most expensive ticket: what were that passenger's sex, age and class?
4. How many 1st-class passengers travelled alone (`alone` column)?

```python
# Practice solutions
# 1
print(df["embark_town"].value_counts())

# 2
print("men:  ", df[df["sex"] == "male"]["survived"].mean())
print("women:", df[df["sex"] == "female"]["survived"].mean())

# 3
print(df.sort_values("fare", ascending=False)[["sex", "age", "pclass", "fare"]].head(1))

# 4 — `alone` is already True/False, so it works as a mask by itself
print(len(df[(df["pclass"] == 1) & df["alone"]]))
```
