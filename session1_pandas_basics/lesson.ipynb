## Part 1 — DataFrame & Series

A **DataFrame** is a whole table (rows and columns). A **Series** is a single column.
We import pandas as `pd` and seaborn as `sns` — these short names are a convention everyone uses.

```python
import pandas as pd
import seaborn as sns
```

seaborn comes with small practice datasets. `sns.load_dataset("titanic")` downloads one and returns a DataFrame.
With your own data you would write `pd.read_csv("file.csv")` — the result is the same kind of DataFrame.

```python
df = sns.load_dataset("titanic")
df
```

`head()` shows the first 5 rows, `tail(n)` the last *n* rows — the quickest way to see what the data looks like.

```python
df.head()
```

```python
df.tail(3)
```

`shape` gives `(rows, columns)`. Note: `shape` is an **attribute** (no brackets), while `head()` is a **method** (needs brackets).
`columns` lists the column names.

```python
df.shape
```

```python
df.columns
```

`info()` is the X-ray of a DataFrame: every column, how many **non-null** (non-missing) values it has, and its data type.
Notice that `age` and `deck` have fewer than 891 non-null values — they contain **missing data** (we deal with that in session 2).

```python
df.info()
```

`describe()` gives quick statistics for every numeric column: count, mean, min, max and quartiles.
Look at it: what is the average age? What is the highest fare? Does anything look strange?

```python
df.describe()
```

### One column = a Series

`df["age"]` selects one column. A Series has its own methods like `mean()`, `min()`, `max()`, `median()`.

```python
ages = df["age"]
type(ages)
```

```python
print(ages.mean())
print(ages.min(), ages.max())
print(ages.median())
```

To select **several columns**, pass a *list* of names — notice the double brackets. The result is a DataFrame, not a Series.

```python
df[["sex", "age", "survived"]].head()
```

`unique()` shows the distinct values in a column, `nunique()` counts them.

```python
df["sex"].unique()
```

```python
df["sex"].nunique()
```

## Part 2 — selecting rows, filtering, sorting

There are two ways to pick rows by their place in the table:
- `iloc` — by **position**, like list indexing (the end is *excluded*)
- `loc` — by **label** (the end is *included*), and you can name the columns

In practice you will use **filtering** much more often than either of them.

```python
df.iloc[0]   # the first row
```

```python
df.iloc[0:5, 0:4]   # rows 0–4 and the first 4 columns (end excluded)
```

```python
df.loc[0:4, ["sex", "age", "fare"]]   # rows labelled 0 to 4 (end included), columns by name
```

### Filtering with a boolean mask

A condition on a column gives a Series of `True`/`False` — called a **mask**.
`df[mask]` keeps only the rows where the mask is `True`.

```python
mask = df["age"] > 60
mask.head(10)
```

```python
df[mask]
```

```python
len(df[mask])   # how many passengers were over 60
```

Combine conditions with `&` (and) and `|` (or).
**Each condition must be in its own parentheses** — forgetting them is the most common beginner error.
`isin([...])` checks if a value is one of several options.

```python
first_class_women = df[(df["sex"] == "female") & (df["pclass"] == 1)]
first_class_women.shape
```

```python
# filter the rows, then keep only 3 columns
df[(df["age"] < 10) | (df["age"] > 70)][["age", "sex", "survived"]]
```

```python
df[df["embark_town"].isin(["Cherbourg", "Queenstown"])].shape
```

### Sorting & counting

`sort_values(col)` sorts the rows; add `ascending=False` for largest first.
`value_counts()` counts how often each value appears; `normalize=True` gives proportions instead.

```python
df.sort_values("fare", ascending=False).head()
```

```python
df["pclass"].value_counts()
```

```python
df["pclass"].value_counts(normalize=True)
```

**Useful trick:** `survived` is `0` or `1`, so its **mean is the survival rate** (0.38 = 38% survived).
You will use this idea all through the course.

```python
df["survived"].mean()
```

```python
# survival rate of women: filter first, then take the mean
df[df["sex"] == "female"]["survived"].mean()
```
