## Part 1 — values, types, comparisons

A **variable** is a name for a value. Every value has a **type**:
`int` (whole number), `float` (decimal number), `str` (text), `bool` (`True`/`False`).
`type()` tells you the type. In pandas, every column has a type too (you'll see `int64`, `float64`, `object`/`str`).

```python
age = 29
fare = 71.28
name = "Mary"
survived = True

print(type(age), type(fare), type(name), type(survived))
```

Converting between types: `int()`, `float()`, `str()`. Note that `int()` cuts off the decimals, it doesn't round.

```python
print(int(7.9))        # 7 — decimals are cut off
print(round(7.9))      # 8 — rounded
print(float("3.5"))    # text -> number
print(str(42) + "!")   # number -> text, so it can be joined to other text
```

**f-strings** put values inside text: write `f` before the quotes and the variable in `{}`.
`{x:.2f}` shows 2 decimals, `{x:.1%}` shows a percentage.

```python
rate = 0.3838
print(f"{name} is {age} years old and paid {fare:.2f}")
print(f"Survival rate: {rate:.1%}")
```

**Comparisons** give a `bool`: `==` (equal), `!=` (not equal), `>`, `<`, `>=`, `<=`.
Combine them with `and`, `or`, `not`.
This is exactly how you will filter rows in pandas (there, with `&` and `|` instead of `and`/`or`).

```python
print(age > 18)
print(name == "Mary")
print(age > 18 and fare > 50)   # both must be True
print(age < 12 or age > 60)     # at least one must be True
```

`None` means "no value". In pandas, missing values usually appear as `NaN` (not a number) — same idea.

```python
cabin = None
print(cabin is None)
```

## Part 2 — lists and dictionaries

A **list** is an ordered collection in square brackets. Positions start at **0**; negative positions count from the end.

```python
ages = [22, 38, 26, 35, 54, 2, 27]
print(ages[0], ages[-1], len(ages))
```

**Slicing** `list[start:end]` takes a range — the end is *excluded*. pandas `iloc` works the same way.

```python
print(ages[0:3])   # first three
print(ages[:2])    # from the start
print(ages[4:])    # until the end
```

Lists have useful methods and built-in functions: `append`, `sorted`, `sum`, `min`, `max`.
`in` checks whether a value is in the list.

```python
ages.append(80)                # add a value at the end
print(sorted(ages))
print(sum(ages) / len(ages))   # the average
print(min(ages), max(ages))
print(35 in ages)
```

A **dictionary** maps **keys** to **values**, in curly braces. Look a value up with its key.
You'll use dicts in pandas for `rename(columns={"old": "new"})` — and a DataFrame itself is like a dict of columns.

```python
passenger = {"name": "Mary", "age": 29, "class": 1}
print(passenger["age"])

passenger["survived"] = True        # add a key
print(passenger.keys())
print(passenger.get("cabin", "unknown"))   # default if the key is missing
```

A list of dicts is a natural way to hold table-like data — pandas can turn it straight into a DataFrame (session 1).

```python
people = [
    {"name": "Mary", "age": 29},
    {"name": "John", "age": 41},
]
print(people[1]["name"])
```

## Part 3 — if, loops, comprehensions

`if / elif / else` runs code depending on a condition. **Indentation** (4 spaces) shows which lines belong to the block.

```python
age = 15
if age < 12:
    group = "child"
elif age < 18:
    group = "teen"
else:
    group = "adult"
print(group)
```

A `for` loop runs code once for every item. The indented lines are repeated; here an `if` inside the loop decides what to print.

```python
for a in [5, 30, 70]:
    if a > 60:
        print(a, "is old")
    else:
        print(a, "is not old")
```

`enumerate` gives you the position too; `dict.items()` loops over key–value pairs.

```python
for i, town in enumerate(["Southampton", "Cherbourg", "Queenstown"]):
    print(i, town)

for key, value in passenger.items():
    print(f"{key}: {value}")
```

A **list comprehension** builds a new list in one line: `[expression for item in list if condition]`.

**Good to know:** in pandas you almost never write loops — you do math on whole columns at once.
But comprehensions are handy for preparing lists, e.g. of column names.

```python
ages = [22, 38, 26, 35, 54, 2, 27]
print([a * 12 for a in ages])        # ages in months
print([a for a in ages if a < 18])   # only children
```

## Part 4 — functions, lambda, imports, methods

A **function** packs code you want to reuse. `def` defines it, `return` gives back the result.
Arguments can have **default values**.

```python
def age_group(age, adult_age=18):
    if age < adult_age:
        return "child"
    return "adult"

print(age_group(10), age_group(40), age_group(19, adult_age=21))
```

A **lambda** is a tiny function without a name, written in one line: `lambda x: x * 2`.
You'll pass lambdas to pandas, for example `df["age"].apply(lambda a: a < 18)` in session 2.

A common use is `key=`: `sorted(people, key=...)` sorts by whatever the function returns —
below, by each person's age. `reverse=True` puts the largest first.

```python
double = lambda x: x * 2
print(double(5))
print(sorted(people, key=lambda p: p["age"], reverse=True))
```

**Libraries** add extra tools. `import ... as ...` gives them a short name — `pd`, `sns`, `plt`, `px` are the standard ones you'll use.

```python
import math
import statistics as st

print(math.sqrt(16), st.median(ages))
```

**Methods vs. attributes** — both are accessed with a dot:
- a **method** *does* something and needs brackets: `text.upper()`, later `df.head()`
- an **attribute** is just a stored value, no brackets: later `df.shape`, `df.columns`

Methods can be **chained**: each one works on the result of the previous one. pandas code uses this a lot.

```python
text = "  Southampton  "
print(text.strip().lower().replace("south", "SOUTH"))
```

**Errors** are normal — read the *last line* of the message first: it says what went wrong.
Common ones: `NameError` (typo in a name / cell not run yet), `KeyError` (key or column doesn't exist),
`TypeError` (wrong type, like adding text to a number), `IndentationError`.

`try` / `except` catches an error so the program can keep going. Below we use it to look at a `KeyError` without stopping the notebook.

```python
try:
    passenger["cabin"]
except KeyError as e:
    print("KeyError:", e)
```
