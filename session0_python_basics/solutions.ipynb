## Exercises

1. Create variables for a passenger: name, age, fare. Print `"<name> (<age>) paid <fare>"` with the fare to 1 decimal.
2. Given `fares = [7.25, 71.28, 7.92, 53.1, 26.0, 8.05, 13.0, 30.07]`: print the average, the 3 most expensive, and how many are below 10.
3. Write a function `fare_class(fare)` that returns `"cheap"` (< 10), `"normal"` (< 50) or `"expensive"`. Use it on every fare with a list comprehension.
4. Make a dict that counts how many fares fall in each class from exercise 3 (tip: loop and add 1 to `counts[c]`).

```python
# Solution 1
name = "John"
age = 41
fare = 53.1
print(f"{name} ({age}) paid {fare:.1f}")
```

```python
# Solution 2
fares = [7.25, 71.28, 7.92, 53.1, 26.0, 8.05, 13.0, 30.07]
print(sum(fares) / len(fares))             # average
print(sorted(fares, reverse=True)[:3])     # 3 most expensive
print(len([f for f in fares if f < 10]))   # how many are below 10
```

```python
# Solution 3
def fare_class(fare):
    if fare < 10:
        return "cheap"
    elif fare < 50:
        return "normal"
    return "expensive"

classes = [fare_class(f) for f in fares]
classes
```

```python
# Solution 4
counts = {}
for c in classes:
    counts[c] = counts.get(c, 0) + 1
counts
```

## Practice
1. Given `names = ["Mary", "John", "Anna", "Peter"]` and `ages = [29, 41, 8, 63]`, print each person as `"Mary: 29"` (tip: `zip(names, ages)` loops over both lists together).
2. Build a list of dicts `[{"name": ..., "age": ...}, ...]` from the same two lists, then print only the names of people over 18.
3. Write a function `describe(numbers)` that returns a dict with `count`, `mean`, `min` and `max` — a mini version of pandas `describe()`.

```python
# Practice solutions
names = ["Mary", "John", "Anna", "Peter"]
ages = [29, 41, 8, 63]

# 1
for n, a in zip(names, ages):
    print(f"{n}: {a}")

# 2
people = [{"name": n, "age": a} for n, a in zip(names, ages)]
print([p["name"] for p in people if p["age"] > 18])

# 3
def describe(numbers):
    return {
        "count": len(numbers),
        "mean": sum(numbers) / len(numbers),
        "min": min(numbers),
        "max": max(numbers),
    }

describe(ages)
```
