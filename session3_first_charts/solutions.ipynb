```python
# Setup
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
df = sns.load_dataset("titanic")
```

## Exercises
Make a 2×2 figure that tells the survival story, then save it as `titanic_story.png`:
1. Age distribution, split by survival
2. Survival rate per class
3. Survival rate per class and sex
4. Fare per class (box plot)

Give every chart a title. Tip: with 2 rows, the Axes are `axes[0, 0]`, `axes[0, 1]`, `axes[1, 0]`, `axes[1, 1]`.

```python
# Solution
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(data=df, x="age", hue="alive", multiple="stack", bins=20, ax=axes[0, 0])
axes[0, 0].set_title("Age distribution by survival")

sns.barplot(data=df, x="pclass", y="survived", errorbar=None, ax=axes[0, 1])
axes[0, 1].set_title("Survival rate per class")

sns.barplot(data=df, x="pclass", y="survived", hue="sex", errorbar=None, ax=axes[1, 0])
axes[1, 0].set_title("Survival rate per class and sex")

sns.boxplot(data=df, x="pclass", y="fare", ax=axes[1, 1])
axes[1, 1].set_title("Fare per class")

fig.tight_layout()
fig.savefig("titanic_story.png", dpi=150)
plt.show()
```

## Practice
1. Create `age_group` with `pd.cut` (from session 2) and plot the survival rate per age group.
2. Count passengers per `embark_town`, split by class.
3. Under each chart write one sentence: what does it tell us?

```python
# Practice solutions
# 1
df["age_group"] = pd.cut(df["age"], bins=[0, 12, 18, 35, 60, 100],
                         labels=["child", "teen", "young adult", "adult", "senior"])
ax = sns.barplot(data=df, x="age_group", y="survived", errorbar=None)
ax.set_title("Survival rate per age group")
plt.show()

# 2
ax = sns.countplot(data=df, x="embark_town", hue="pclass")
ax.set_title("Passengers per embark town and class")
plt.show()
```
