# Session 3 — first charts (Titanic)

Numbers in a table are hard to read; a good chart shows the pattern instantly.
In this session you make your first charts with **matplotlib** (the base plotting library) and **seaborn** (built on top of it, much shorter code).

**You will learn to:**
- understand the figure/axes model and label a chart
- make histograms, count plots, bar plots and box plots
- put several charts in one figure and save it as an image
- choose the right chart for a question

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")
````

## Part 1 — the mental model

- **Figure** = the whole canvas (the image)
- **Axes** = one chart inside the figure, with its x and y axis

`plt.subplots()` creates both. You draw on the Axes (`ax`) and label it with `set_title`, `set_xlabel`, `set_ylabel`.
`plt.show()` displays the result.

````python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(df["age"].dropna(), bins=20)   # dropna(): skip passengers whose age is missing
ax.set_title("Age of Titanic passengers")
ax.set_xlabel("Age")
ax.set_ylabel("Number of passengers")
plt.show()
````

pandas has quick built-in plots via `.plot(kind=...)` — handy for a fast look while exploring, less good for final charts.

````python
df["age"].plot(kind="hist", bins=20, title="Age (quick look)")
plt.show()
````

````python
df["pclass"].value_counts().sort_index().plot(kind="bar", title="Passengers per class")
plt.show()
````

### seaborn: one line, straight from a DataFrame

Every seaborn chart follows the same pattern:

```python
sns.xxxplot(data=df, x="column", y="column", hue="column")
```

`hue` splits the data by color. We use the `alive` column (yes/no) instead of `survived` (0/1) because it gives a clearer legend.
`sns.set_theme()` sets a clean style for all following charts.

````python
sns.set_theme(style="whitegrid")
````

**Histogram** — how one number is distributed. `multiple="stack"` stacks the colors on top of each other.

````python
sns.histplot(data=df, x="age", bins=20, hue="alive", multiple="stack")
plt.show()
````

**Count plot** — how many rows are in each category.

````python
sns.countplot(data=df, x="pclass", hue="alive")
plt.show()
````

**Bar plot** — the **mean** of `y` per category. With `y="survived"` that's the survival rate.
By default it also draws black lines showing uncertainty; `errorbar=None` hides them.

````python
sns.barplot(data=df, x="pclass", y="survived", hue="sex", errorbar=None)
plt.show()
````

**Box plot** — compares distributions. The box holds the middle 50% of values, the line inside is the median,
and the dots are outliers (unusually high or low values).

````python
sns.boxplot(data=df, x="pclass", y="fare")
plt.show()
````

seaborn functions return the Axes, so you can label them with the same `set_...` methods as before.

````python
ax = sns.barplot(data=df, x="sex", y="survived", errorbar=None)
ax.set_title("Survival rate by sex")
ax.set_xlabel("")
ax.set_ylabel("Survival rate")
plt.show()
````

## Part 2 — several charts, saving, choosing

`plt.subplots(1, 2)` creates a figure with 1 row and 2 Axes. Tell each seaborn chart where to draw with `ax=axes[0]`, `ax=axes[1]`.
`tight_layout()` prevents overlapping labels, and `savefig` saves the figure as an image file.

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.countplot(data=df, x="pclass", hue="alive", ax=axes[0])
axes[0].set_title("Passengers per class")

sns.barplot(data=df, x="pclass", y="survived", errorbar=None, ax=axes[1])
axes[1].set_title("Survival rate per class")

fig.tight_layout()
fig.savefig("titanic_overview.png", dpi=150)
plt.show()
````

### Which chart for which question?

Start from the question, then pick the chart:

| Question | Chart | seaborn |
|---|---|---|
| How is one number distributed? | histogram / box | `histplot`, `boxplot` |
| How many in each category? | count bar | `countplot` |
| Compare an average across categories | bar | `barplot` |
| How does it change over time? | line | `lineplot` (next session) |
| Are two numbers related? | scatter | `scatterplot` |

**Scatter plot** — one dot per row, two numbers on the axes. Is there a relationship between age and fare?
(Not really — the dots don't form a pattern.) `alpha` makes dots transparent so overlaps are visible.

````python
sns.scatterplot(data=df, x="age", y="fare", hue="alive", alpha=0.6)
plt.show()
````
