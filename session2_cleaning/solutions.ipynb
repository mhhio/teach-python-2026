# Session 2 — cleaning & transforming (Titanic) — solutions

Try the exercises in `lesson.ipynb` yourself first. Run the setup cell before the others: it loads the data the solutions use.

```python
# Setup
import pandas as pd
import seaborn as sns

# the clean DataFrame from the lesson — the practice solutions use it
df = sns.load_dataset("titanic")
df = df.drop(columns=["deck"])
df = df.dropna(subset=["embark_town"])
df["age"] = df["age"].fillna(df["age"].median())
df = df[["survived", "pclass", "sex", "age", "sibsp", "parch", "fare", "embark_town"]]
df = df.rename(columns={"sibsp": "siblings_spouses", "parch": "parents_children"})
df["age"] = df["age"].astype(int)
```

```python
# Solution 1
raw = sns.load_dataset("titanic")
raw.isna().sum()
```

```python
# Solution 2
raw["age"] = raw["age"].fillna(raw["age"].median())
raw = raw.dropna(subset=["embark_town"])
raw[["age", "embark_town"]].isna().sum()
```

```python
# Solution 3
raw["family_size"] = raw["sibsp"] + raw["parch"] + 1
raw.groupby("family_size")["survived"].mean().round(2)
```

```python
# Solution 4
raw.groupby(["embark_town", "sex"])["survived"].mean().unstack().round(2)
```

## Practice
1. Add `fare_group` with `pd.qcut(df["fare"], 4, labels=["low", "mid", "high", "very high"])` (4 equal-sized groups). Survival rate per fare group?
2. Average age of survivors vs. non-survivors, per class (tip: `groupby(["pclass", "survived"])`).

```python
# Practice solutions
# 1
df["fare_group"] = pd.qcut(df["fare"], 4, labels=["low", "mid", "high", "very high"])
print(df.groupby("fare_group", observed=True)["survived"].mean().round(2))

# 2
print(df.groupby(["pclass", "survived"])["age"].mean().unstack().round(1))

# 3
df.to_csv("titanic_clean.csv", index=False)
```
